# Flight Operations — Complete Pandas Data Analysis

## Objective
Analyze the Flight Operations dataset using **Pandas** and follow a complete data analysis workflow:

1. Load and understand the dataset
2. Inspect structure, data types, missing values, and statistics
3. Select and filter relevant records
4. Sort records
5. Group and aggregate data
6. Transform data using derived columns
7. Answer meaningful business questions
8. Summarize 5–8 key observations

**Dataset:** `Day10_Flight_Operations_Dataset.csv`

## 1. Import Libraries

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
```

## 2. Load the Dataset

The notebook first checks the common Google Colab location and then the current working directory. If the CSV is uploaded manually to Colab, upload `Day10_Flight_Operations_Dataset.csv` before running this cell.

```python
from pathlib import Path

possible_paths = [
    Path("/content/Flight_Operations.csv"),
    Path("/content/Day10_Flight_Operations_Dataset.csv"),
    Path("Flight_Operations.csv"),
    Path("Day10_Flight_Operations_Dataset.csv")
]

data_path = next((p for p in possible_paths if p.exists()), None)

if data_path is None:
    raise FileNotFoundError(
        "Flight_Operations.csv was not found. Upload the dataset to Google Colab and run this cell again."
    )

df = pd.read_csv(data_path)
print(f"Loaded: {data_path}")
print(f"Shape: {df.shape}")
df.head()
```

## 3. Understand the Dataset

### Basic structure

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())
```

## 4. Check Data Quality

We check duplicate records, missing values, and unique values in categorical columns.

```python
print("Duplicate rows:", df.duplicated().sum())

print("\nMissing values:")
display(df.isna().sum().to_frame("Missing_Count"))

print("\nUnique values in categorical columns:")
for col in df.select_dtypes(include="object").columns:
    print(f"{col}: {df[col].nunique()} unique values")
```

## 5. Descriptive Statistics

`describe()` gives a quick statistical summary of the numerical variables.

```python
display(df.describe())

print("\nFull summary including categorical columns:")
display(df.describe(include="all").T)
```

## 6. Data Type Conversion

`Flight_Date` is converted from text to a proper datetime column so that date based analysis can be performed.

```python
df["Flight_Date"] = pd.to_datetime(df["Flight_Date"], errors="coerce")

print(df["Flight_Date"].dtype)
print("Date range:", df["Flight_Date"].min().date(), "to", df["Flight_Date"].max().date())
```

## 7. Selecting Relevant Data

### Selecting columns

```python
selected_columns = df[
    ["Flight_ID", "Airline", "Origin", "Destination",
     "Passengers", "Seat_Capacity", "Average_Ticket_Price",
     "Delay_Minutes", "Flight_Status", "Passenger_Satisfaction"]
]

display(selected_columns.head(10))
```

### Selecting rows with `loc` and `iloc`

```python
print("First three rows using iloc:")
display(df.iloc[:3])

print("Selected fields for the first five flights using loc:")
display(df.loc[:4, ["Flight_ID", "Airline", "Flight_Status", "Passenger_Satisfaction"]])
```

## 8. Filtering Records

The following questions help identify flights with operational or customer impact characteristics.

```python
# Delayed flights
delayed_flights = df[df["Flight_Status"] == "Delayed"]
print("Number of delayed flights:", len(delayed_flights))
display(delayed_flights.head(10))

# Flights with delay above 30 minutes
delay_above_30 = df[df["Delay_Minutes"] > 30]
print("Flights delayed by more than 30 minutes:", len(delay_above_30))
display(
    delay_above_30[
        ["Flight_ID", "Airline", "Origin", "Destination", "Delay_Minutes", "Weather"]
    ].head(10)
)

# High-satisfaction flights
high_satisfaction = df[df["Passenger_Satisfaction"] >= 4]
print("Flights with satisfaction >= 4:", len(high_satisfaction))
```

## 9. Sorting Records

Sorting can reveal the flights with the highest delays, highest ticket prices, and strongest passenger satisfaction.

```python
print("Top 10 flights by delay:")
display(
    df.sort_values("Delay_Minutes", ascending=False)[
        ["Flight_ID", "Airline", "Origin", "Destination", "Delay_Minutes", "Flight_Status"]
    ].head(10)
)

print("Top 10 ticket prices:")
display(
    df.sort_values("Average_Ticket_Price", ascending=False)[
        ["Flight_ID", "Airline", "Travel_Class", "Average_Ticket_Price"]
    ].head(10)
)

print("Top 10 satisfaction scores:")
display(
    df.sort_values("Passenger_Satisfaction", ascending=False)[
        ["Flight_ID", "Airline", "Passenger_Satisfaction", "Flight_Status"]
    ].head(10)
)
```

## 10. Grouping and Aggregation

### Flights by airline

We compare flight count, passengers, average delay, average ticket price, and satisfaction across airlines.

```python
airline_summary = (
    df.groupby("Airline")
      .agg(
          Flights=("Flight_ID", "count"),
          Total_Passengers=("Passengers", "sum"),
          Avg_Delay_Minutes=("Delay_Minutes", "mean"),
          Avg_Ticket_Price=("Average_Ticket_Price", "mean"),
          Avg_Satisfaction=("Passenger_Satisfaction", "mean")
      )
      .sort_values("Flights", ascending=False)
)

display(airline_summary.round(2))
```

### Flight status analysis

```python
status_summary = (
    df.groupby("Flight_Status")
      .agg(
          Flights=("Flight_ID", "count"),
          Avg_Delay_Minutes=("Delay_Minutes", "mean"),
          Avg_Satisfaction=("Passenger_Satisfaction", "mean")
      )
      .sort_values("Flights", ascending=False)
)

display(status_summary.round(2))
```

### Weather and delay

```python
weather_summary = (
    df.groupby("Weather")
      .agg(
          Flights=("Flight_ID", "count"),
          Avg_Delay_Minutes=("Delay_Minutes", "mean"),
          Avg_Satisfaction=("Passenger_Satisfaction", "mean")
      )
      .sort_values("Avg_Delay_Minutes", ascending=False)
)

display(weather_summary.round(2))
```

### Travel class

```python
class_summary = (
    df.groupby("Travel_Class")
      .agg(
          Flights=("Flight_ID", "count"),
          Avg_Ticket_Price=("Average_Ticket_Price", "mean"),
          Avg_Passengers=("Passengers", "mean"),
          Avg_Satisfaction=("Passenger_Satisfaction", "mean")
      )
      .sort_values("Avg_Ticket_Price", ascending=False)
)

display(class_summary.round(2))
```

## 11. Data Transformation

New columns are created using Pandas operations:

- **Load_Factor_%** = passengers ÷ seat capacity × 100
- **Estimated_Revenue** = passengers × average ticket price
- **Delay_Category** = categorical classification of delay duration
- **Month** = month extracted from the flight date

```python
analysis_df = df.copy()

analysis_df["Load_Factor_%"] = (
    analysis_df["Passengers"] / analysis_df["Seat_Capacity"] * 100
).round(2)

analysis_df["Estimated_Revenue"] = (
    analysis_df["Passengers"] * analysis_df["Average_Ticket_Price"]
)

analysis_df["Delay_Category"] = pd.cut(
    analysis_df["Delay_Minutes"],
    bins=[-1, 0, 15, 60, np.inf],
    labels=[
        "No Delay",
        "Minor (1-15 min)",
        "Moderate (16-60 min)",
        "Severe (>60 min)"
    ]
)

analysis_df["Month"] = analysis_df["Flight_Date"].dt.month_name()

display(
    analysis_df[
        [
            "Flight_ID", "Airline", "Passengers", "Seat_Capacity",
            "Load_Factor_%", "Average_Ticket_Price",
            "Estimated_Revenue", "Delay_Minutes", "Delay_Category", "Month"
        ]
    ].head(10)
)
```

## 12. Load Factor Analysis

A load factor above 100% is not operationally realistic for a standard passenger capacity measure. If it occurs in the dataset, it should be treated as a **data quality finding**, not automatically corrected.

```python
load_factor_summary = (
    analysis_df.groupby("Airline")
    .agg(
        Flights=("Flight_ID", "count"),
        Avg_Load_Factor=("Load_Factor_%", "mean"),
        Max_Load_Factor=("Load_Factor_%", "max")
    )
    .sort_values("Avg_Load_Factor", ascending=False)
)

display(load_factor_summary.round(2))

print("Flights with load factor above 100%:", (analysis_df["Load_Factor_%"] > 100).sum())
```

## 13. Estimated Revenue by Airline

This is a simple analytical estimate based on passengers multiplied by average ticket price. It is **not accounting revenue**, because the dataset does not provide actual transaction level revenue.

```python
revenue_by_airline = (
    analysis_df.groupby("Airline")
    .agg(
        Estimated_Revenue=("Estimated_Revenue", "sum"),
        Total_Passengers=("Passengers", "sum"),
        Avg_Ticket_Price=("Average_Ticket_Price", "mean")
    )
    .sort_values("Estimated_Revenue", ascending=False)
)

display(revenue_by_airline.round(2))
```

## 14. Route-Level Analysis

Routes with frequent flights and routes with high average delays can be identified using multi column grouping.

```python
route_summary = (
    analysis_df.groupby(["Origin", "Destination"])
    .agg(
        Flights=("Flight_ID", "count"),
        Total_Passengers=("Passengers", "sum"),
        Avg_Delay_Minutes=("Delay_Minutes", "mean"),
        Avg_Satisfaction=("Passenger_Satisfaction", "mean")
    )
    .sort_values("Flights", ascending=False)
)

print("Most frequently operated routes:")
display(route_summary.head(10).round(2))

print("Routes with highest average delay among routes with at least 5 flights:")
display(
    route_summary[route_summary["Flights"] >= 5]
    .sort_values("Avg_Delay_Minutes", ascending=False)
    .head(10)
    .round(2)
)
```

## 15. Monthly Analysis

The transformed `Month` column allows us to compare operations and customer satisfaction across months.

```python
monthly_summary = (
    analysis_df.groupby("Month", sort=False)
    .agg(
        Flights=("Flight_ID", "count"),
        Avg_Delay_Minutes=("Delay_Minutes", "mean"),
        Avg_Satisfaction=("Passenger_Satisfaction", "mean"),
        Total_Passengers=("Passengers", "sum")
    )
)

month_order = [
    "January", "February", "March",
    "April", "May", "June"
]

monthly_summary = monthly_summary.reindex(month_order)
display(monthly_summary.round(2))
```

## 16. Booking Channel Analysis

```python
booking_summary = (
    analysis_df.groupby("Booking_Channel")
    .agg(
        Flights=("Flight_ID", "count"),
        Avg_Ticket_Price=("Average_Ticket_Price", "mean"),
        Avg_Satisfaction=("Passenger_Satisfaction", "mean")
    )
    .sort_values("Flights", ascending=False)
)

display(booking_summary.round(2))
```

## 17. Overall KPI Snapshot

```python
kpis = pd.Series({
    "Total Flights": len(analysis_df),
    "Total Passengers": analysis_df["Passengers"].sum(),
    "On-Time Flights": (analysis_df["Flight_Status"] == "On Time").sum(),
    "Delayed Flights": (analysis_df["Flight_Status"] == "Delayed").sum(),
    "Cancelled Flights": (analysis_df["Flight_Status"] == "Cancelled").sum(),
    "Average Delay (min)": analysis_df["Delay_Minutes"].mean(),
    "Average Satisfaction": analysis_df["Passenger_Satisfaction"].mean(),
    "Average Load Factor (%)": analysis_df["Load_Factor_%"].mean(),
    "Estimated Revenue": analysis_df["Estimated_Revenue"].sum()
})

display(kpis.to_frame("Value"))
```

## 18. Key Observations

Based on the analysis of the supplied dataset:

1. **On time performance dominates:** 114 of 180 flights are marked On Time, while 59 are Delayed and 7 are Cancelled.
2. **Delays affect satisfaction:** delayed flights have an average passenger satisfaction of about **3.27/5**, compared with about **3.59/5** for on time flights.
3. **Rain is associated with the highest average delay:** among the weather categories, Rain has an average delay of about **22.22 minutes**, followed by Storm at about 20.10 minutes.
4. **Vistara has the highest estimated revenue:** using the simple passengers × average ticket price calculation, Vistara generates the largest estimated revenue in this dataset.
5. **Premium Economy has the highest satisfaction among travel classes:** its average satisfaction is about **3.69/5**, higher than Economy and Business.
6. **April shows the highest average delay:** April has an average delay of about **25.41 minutes**, while March has the lowest at about **10.69 minutes**.
7. **There are missing delay values:** all 7 missing `Delay_Minutes` records correspond to Cancelled flights. These should not automatically be replaced with zero because a cancelled flight does not necessarily mean zero delay.
8. **The dataset contains load factor anomalies:** several flights have more passengers than recorded seat capacity, producing load factors above 100%. This is an important data quality issue that should be investigated before using load factor for operational decisions.

## 19. Conclusion

The Pandas workflow shows how a raw flight operations dataset can be converted into useful operational insights. The analysis covered data loading, inspection, descriptive statistics, selection, filtering, sorting, grouping, aggregation, and transformation.

The strongest patterns are related to **flight status, delays, weather, travel class, monthly performance, revenue estimates, and data quality**. The analysis also demonstrates why missing values and unrealistic derived metrics should be investigated before drawing business conclusions.
